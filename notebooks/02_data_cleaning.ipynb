Before working with the dataset, we import the Python libraries required for data cleaning.

- **Pandas** is used for loading, cleaning, and manipulating tabular data.

```python
# ============================================================
# Kenya Retail Sales Analysis
# Notebook 02 - Data Cleaning & Feature Engineering
# ============================================================
import pandas as pd
import numpy as np
```

Although the dataset was explored in Notebook 1, each notebook should be self-contained. Therefore, I load the dataset again so that this notebook can run independently.

```python
df = pd.read_csv("../data/Kenya_Retail_Sales_Dataset.csv")
df.head()
```

## Step 3: Inspect Current Data Types

Before cleaning the dataset, it is important to verify the data type of every column.

Incorrect data types can lead to inaccurate calculations and prevent time-based analysis.

For example:

- Dates stored as text cannot be grouped by month or year.
- Times stored as text cannot be used to identify peak shopping hours.

Understanding the current structure of the dataset helps determine the cleaning steps required.

```python
df.info()
```

# Step 4: Convert the Date Column

The **Date** column is currently stored as text.

Converting it to a datetime format enables powerful time-based analysis such as:

- yearly sales
- monthly trends
- quarterly performance
- weekday analysis

Without this conversion, Pandas cannot perform date calculations.

```python
df["Date"] = pd.to_datetime(df["Date"])
df["Date"].head()
```

## Verify the Date Conversion

After converting the Date column, we verify that its data type has changed from **object** to **datetime64**.

This confirms that the conversion was successful.

```python
df.info()
```

# Step 5: Convert the Time Column

The Time column is also stored as text.

Converting it allows us to identify:

- peak shopping hours
- customer purchasing patterns
- busy periods throughout the day

This information can support staffing decisions and promotional planning.

```python
df["Time"].head()
```

```python
df["Time"] = pd.to_datetime(df["Time"], format="%H:%M:%S")
```

```python
df["Time"].dtype
```

```python
df["Time"] = pd.to_datetime(df["Time"], format="%H:%M:%S").dt.time
```

```python
df["Time"].head()
```

# Step 6: Feature Engineering

Feature engineering is the process of creating new variables from existing data.

Rather than repeatedly extracting information from the Date column during analysis, we create dedicated columns that can be used directly in SQL queries, Power BI dashboards, and visualizations.

These additional features improve efficiency and make business analysis much easier.
## Create the Year Column

The Year column enables annual performance analysis.

Example business question:

**How has revenue changed over the years?**

```python
df["Year"] = df["Date"].dt.year 
df["Month"] = df["Date"].dt.month   
```

## Create the Month Name Column

Month names improve the readability of reports and dashboards.

Instead of displaying numbers such as 1, 2, and 3, dashboards will display January, February, and March.

```python
df["Month Name"] = df["Date"].dt.month_name()
```

## Create the Quarter Column

Many organizations evaluate business performance quarterly.

Creating this feature allows us to perform quarterly sales analysis and prepare executive reports.

```python
df["Quarter"] = df["Date"].dt.quarter 
```

## Create the Day Column

The Day column stores the day of the month.

This feature is useful when analysing daily sales performance.
## Create the Day Name Column

The Day Name column helps identify customer shopping behaviour across different days of the week.

Example business question:

**Which day records the highest sales?**

```python
df["Day"]= df["Date"].dt.day
df["Day Name"] = df["Date"].dt.day_name()
```

## Create the Hour Column

The Hour column is extracted from the transaction time.

This feature helps determine peak shopping hours and customer purchasing behaviour throughout the day.

```python
df["Hour"] = df["Time"].dt.hour
```

## Classify Transactions as Weekday or Weekend

Customer shopping habits often differ between weekdays and weekends.

Creating this feature allows us to compare sales performance between these two periods.

```python
df["Day Type"] = np.where(
    df["Date"].dt.dayofweek < 5,
    "Weekday",
    "Weekend"
)
```

```python
# ==========================================================
# Step 9: Validate Numerical Columns
# ==========================================================

print("=" * 60)
print("VALIDATING NUMERICAL COLUMNS")
print("=" * 60)

# Display summary statistics
print("\nSummary Statistics")
print("-" * 60)
display(df.describe())

# Check for invalid Quantity values
invalid_quantity = df[df["Quantity"] <= 0]
print(f"\nInvalid Quantity Records: {len(invalid_quantity)}")

# Check for invalid Unit Price values
invalid_price = df[df["Unit_Price"] <= 0]
print(f"Invalid Unit Price Records: {len(invalid_price)}")

# Check for invalid Total Sales values
invalid_sales = df[df["Total_Sales"] <= 0]
print(f"Invalid Total Sales Records: {len(invalid_sales)}")

# Check for invalid Tax values
invalid_tax = df[df["Tax"] < 0]
print(f"Invalid Tax Records: {len(invalid_tax)}")

# Check for invalid Rating values
invalid_rating = df[(df["Rating"] < 0) | (df["Rating"] > 10)]
print(f"Invalid Rating Records: {len(invalid_rating)}")

# Check for negative Profit values
negative_profit = df[df["Profit"] < 0]
print(f"Negative Profit Records: {len(negative_profit)}")

print("\n" + "=" * 60)

# Final conclusion
if (
    len(invalid_quantity) == 0 and
    len(invalid_price) == 0 and
    len(invalid_sales) == 0 and
    len(invalid_tax) == 0 and
    len(invalid_rating) == 0
):
    print("✅ Data Validation Successful!")
    print("No invalid numerical values were found.")
else:
    print("⚠️ Some numerical columns contain invalid values.")
    print("Review the records above before proceeding.")
```

```python
Q1 = df["Total_Sales"].quantile(0.25)
Q3 = df["Total_Sales"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - (1.5 * IQR)
upper = Q3 + (1.5 * IQR)

outliers = df[
    (df["Total_Sales"] < lower) |
    (df["Total_Sales"] > upper)
]

print(f"Number of potential outliers: {len(outliers)}")
```
