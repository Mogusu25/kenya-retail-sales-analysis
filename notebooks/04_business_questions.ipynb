# Business Questions & Decision Analysis

## Objective

This notebook uses the cleaned Kenya retail sales dataset to answer key business questions related to growth, product strategy, customers, revenue optimization, employee performance, and operational efficiency.

The analysis focuses on converting data insights into practical recommendations that can support management decision-making.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

#Load the cleaned dataset
df = pd.read_csv("../data/processed/Kenya_Retail_Sales_Cleaned.csv")

#Convert the date columns

df["Date"]=pd.to_datetime(df["Date"])

#preview the dataset
print(df.head())
```

## Business Question 1: Which branch should receive more investment?

Branch performance is evaluated using revenue, profit, transaction volume, and average transaction value.

A strong investment candidate should demonstrate both strong sales and profitability rather than relying on revenue alone.

```python
branch_performance = (
    df.groupby("Branch")
    .agg(
        Revenue=("Total_Sales", "sum"),
        Profit=("Profit", "sum"),
        Transactions=("Invoice_ID", "nunique"),
        Average_Transaction=("Total_Sales", "mean")
    )
    .sort_values("Profit", ascending=False)
)

branch_performance
```

```python
branch_performance["Profit_Margin_%"] = (
    branch_performance["Profit"] /
    branch_performance["Revenue"] * 100
).round(2)

branch_performance
```

## Business Question 2: Which city should be prioritized for expansion?

City-level performance is analyzed to identify markets with strong revenue and profit potential.

Expansion should prioritize locations that demonstrate sustainable profitability rather than high sales volume alone.

```python
city_performance = (
    df.groupby("City")
    .agg(
        Revenue=("Total_Sales", "sum"),
        Profit=("Profit", "sum"),
        Transactions=("Invoice_ID", "nunique")
    )
    .sort_values("Profit", ascending=False)
)

city_performance["Profit_Margin_%"] = (
    city_performance["Profit"] /
    city_performance["Revenue"] * 100
).round(2)

city_performance
```

## Business Question 3: Which products should be promoted?

Products are evaluated based on revenue, profit, quantity sold, and profit margin.

Products with strong profitability and sales volume are potential candidates for promotional campaigns and increased visibility.

```python
product_performance = (
    df.groupby("Product")
    .agg(
        Revenue=("Total_Sales", "sum"),
        Profit=("Profit", "sum"),
        Quantity_Sold=("Quantity", "sum"),
        Transactions=("Invoice_ID", "nunique")
    )
    .sort_values("Profit", ascending=False)
)

product_performance["Profit_Margin_%"] = (
    product_performance["Profit"] /
    product_performance["Revenue"] * 100
).round(2)

product_performance.head(10)
```

```python
#Which products underperfomed
product_performance.sort_values(
    ["Profit", "Revenue"],
    ascending=[True, True]
).head(10)
```

```python
#WHICH CUSTOMER STRATEGIES ARE MORE VALUABLE
customer_performance = (
    df.groupby("Customer_Type")
    .agg(
        Revenue=("Total_Sales", "sum"),
        Profit=("Profit", "sum"),
        Transactions=("Invoice_ID", "nunique"),
        Average_Transaction=("Total_Sales", "mean")
    )
)

customer_performance["Profit_Margin_%"] = (
    customer_performance["Profit"] /
    customer_performance["Revenue"] * 100
).round(2)

customer_performance.sort_values("Profit", ascending=False)
```

```python
#How can customer spending be increased?
customer_spending = (
    df.groupby("Customer_Type")
    .agg(
        Average_Transaction=("Total_Sales", "mean"),
        Average_Quantity=("Quantity", "mean"),
        Revenue=("Total_Sales", "sum"),
        Transactions=("Invoice_ID", "nunique")
    )
    .sort_values("Average_Transaction", ascending=False)
)

customer_spending
```

```python
#Which payment methods should be encouraged?
payment_performance = (
    df.groupby("Payment_Method")
    .agg(
        Transactions=("Invoice_ID", "nunique"),
        Revenue=("Total_Sales", "sum"),
        Profit=("Profit", "sum")
    )
    .sort_values("Revenue", ascending=False)
)

payment_performance
```

```python
#WHEN SHOULD PROMOTIONS BE PROMOTED
monthly_performance = (
    df.groupby(["Year", "Month", "Month Name"])
    .agg(
        Revenue=("Total_Sales", "sum"),
        Profit=("Profit", "sum"),
        Transactions=("Invoice_ID", "nunique")
    )
    .reset_index()
    .sort_values(["Year", "Month"])
)

monthly_performance
```

```python
monthly_performance.sort_values(
    "Revenue",
    ascending=False
).head(10)
```

```python
#EMPLOYEES WHO DESERVE RECOGNITION
salesperson_performance = (
    df.groupby("Salesperson")
    .agg(
        Revenue=("Total_Sales", "sum"),
        Profit=("Profit", "sum"),
        Transactions=("Invoice_ID", "nunique")
    )
    .sort_values("Profit", ascending=False)
)

salesperson_performance
```

```python
salesperson_performance.sort_values(
    "Profit",
    ascending=True
).head(10)
```

```python
hourly_activity = (
    df.groupby("Hour")
    .agg(
        Transactions=("Invoice_ID", "nunique"),
        Revenue=("Total_Sales", "sum"),
        Profit=("Profit", "sum")
    )
    .sort_values("Transactions", ascending=False)
)

hourly_activity
```

```python
hourly_activity.head(10)
```

```python
day_type_performance = (
    df.groupby("Day Type")
    .agg(
        Revenue=("Total_Sales", "sum"),
        Profit=("Profit", "sum"),
        Transactions=("Invoice_ID", "nunique")
    )
)

day_type_performance["Average_Transaction"] = (
    day_type_performance["Revenue"] /
    day_type_performance["Transactions"]
).round(2)

day_type_performance
```

```python
#Business Question 13: Which products generate revenue but limited profit?

product_strategy = (
    df.groupby("Product")
    .agg(
        Revenue=("Total_Sales", "sum"),
        Profit=("Profit", "sum"),
        Quantity=("Quantity", "sum")
    )
)

product_strategy["Profit_Margin_%"] = (
    product_strategy["Profit"] /
    product_strategy["Revenue"] * 100
)

product_strategy.sort_values(
    "Revenue",
    ascending=False
).head(15)
```
